# Week 1 – Getting set up, Jupyter, and just enough Python

**What the lab wanted:** create an Anaconda environment, install numpy/matplotlib/Jupyter, open a notebook and run some code.

**What we're doing instead:** exactly the same thing, but with VS Code and a normal Python *virtual environment* (`venv`). Anaconda is just one way of doing this; a `venv` does the same job and is built into Python.

### The big idea: what is an "environment"?
Python on its own can't do much AI. You add **packages** (libraries other people wrote) like `numpy`, `matplotlib` and `torch`.
An **environment** is a folder that holds one copy of Python plus the packages for one project, so different projects don't fight over versions.

| Anaconda word | VS Code / venv equivalent |
|---|---|
| `conda create -n ai python=3.11` | `python -m venv .venv` |
| `conda activate ai` | VS Code does it for you when you pick the `.venv` kernel |
| `conda install numpy` | `pip install numpy` |
| `jupyter notebook` (launch server) | Just open the `.ipynb` file in VS Code |

### The other big idea: what is a "kernel"?
When you run a cell, VS Code sends the code to a running Python process called the **kernel**. The kernel *remembers* everything you've run so far (variables, imports) until you restart it.
In the top-right of this notebook there's a **Select Kernel** button – it must say `.venv` (the environment you made). If it says anything else, click it and choose `.venv`.

> **How to run a cell:** click in it and press **Shift + Enter**. Run this notebook top to bottom.

## 2. How notebooks behave (the bit that confuses everyone)
A notebook is a list of **cells**. There are two kinds:
* **Code cells** – Python that runs when you press Shift + Enter.
* **Markdown cells** – formatted text, like this one. (Double-click one to see the raw text; Shift + Enter to render it again.)

The number on the left of a code cell, e.g. `[3]`, is the **order it was run in**, not its position. The kernel only knows about cells *you have actually run*.

````python
a = 10
print("a is", a)
````

````python
a = a + 5
print("a is now", a)
````

Run the second cell above a few more times – `a` keeps growing (20, 25, ...) because the kernel remembers the old value.

**This is the #1 cause of confusing errors in labs.** If things get weird: click **Restart** in the toolbar, then **Run All**. That re-runs everything top to bottom from a clean state.

### 3.1 Variables, lists and loops

````python
scores = [2.1, 0.5, -1.3]       # a list
print(scores[0])               # indexing starts at 0  -> 2.1
print(len(scores))             # how many items        -> 3

for s in scores:               # loop over every item
    print("score:", s)

for i in range(3):             # range(3) gives 0, 1, 2
    print("i =", i)
````

### 3.2 f-strings (how the labs print results)
Put an `f` before the quotes and anything inside `{ }` is filled in. `:.4f` means "4 decimal places", `:02d` means "integer, padded to 2 digits".

````python
epoch, loss, acc = 3, 0.978812, 0.8449
print(f"Epoch {epoch:02d} | Loss: {loss:.4f} | Test Acc: {acc:.4f}")
````

### 3.3 Functions
`def` makes a reusable block of code. `return` hands a value back. In Week 4 `train(...)` and `evaluate(...)` are just functions like this.

````python
def average(values):
    total = 0
    for v in values:
        total += v          # same as total = total + v
    return total / len(values)

print(average([1, 2, 3, 4]))
````

### 3.4 Classes – the one you *really* need for PyTorch
In Week 4 the neural network is written as a **class**. A class is a blueprint; an **object** is one thing built from that blueprint.

* `__init__` runs once when the object is created – it's where you set things up.
* `self` means "this particular object". `self.name = ...` stores something on the object.
* Other functions inside the class are called **methods**.

````python
class Dog:
    def __init__(self, name):       # runs when you write Dog("Rex")
        self.name = name            # store the name on this dog

    def speak(self):                # a method
        return f"{self.name} says woof"

rex = Dog("Rex")                     # create an object
print(rex.speak())
print(rex.name)
````

**Inheritance:** `class Puppy(Dog):` means "a Puppy is a Dog, plus extra stuff". `super().__init__(...)` runs the parent's setup first.

That's *exactly* the pattern in Week 4:
```python
class NeuralNetwork(nn.Module):      # our network IS a PyTorch Module
    def __init__(self):
        super().__init__()           # let PyTorch do its own setup first
        self.fc1 = nn.Linear(784, 256)   # store a layer on the object
    def forward(self, x):            # a method: how data flows through
        ...
```

````python
class Puppy(Dog):
    def __init__(self, name, age_weeks):
        super().__init__(name)          # run Dog's __init__ (sets self.name)
        self.age_weeks = age_weeks

    def speak(self):                    # replace the parent's method
        return f"{self.name} (age {self.age_weeks} weeks) says yip"

p = Puppy("Bean", 9)
print(p.speak())
````
